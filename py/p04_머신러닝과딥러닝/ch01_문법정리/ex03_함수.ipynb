```python
def calculator(x, y, op):
    err = 0
    result = 0

    try:
        if op == '+':
            result = x + y
        elif op == '-':
            result = x - y  
        elif op == '*':
            result = x * y
        elif op == '/':
            # if y != 0:
                result = x / y
        else:
            err = -1
            print("계산 실패") 
            
        return err, result    
    except:
        err = -1
        print("0으로 나눌 수 없습니다")
        return err, result
```

```python
calculator(10, 3, '/')
```

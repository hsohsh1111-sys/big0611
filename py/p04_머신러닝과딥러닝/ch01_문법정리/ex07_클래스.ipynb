```python
# calculator.py
result = 0

def add(num):
    global result  # 전역변수 result 사용
    result = result + num  # 결괏값(result)에 입력값(num) 더하기
    return result  # 결괏값 반환

print(add(3))
print(add(4))
```

```python
# calculator2.py
result1 = 0
result2 = 0

def add1(num):  # 계산기1
    global result1
    result1 += num
    return result1

def add2(num):  # 계산기2
    global result2
    result2 += num
    return result2

print(add1(3))
print(add1(4))
print(add2(3))
print(add2(7))
```

```python
class FourCal:
    def __init__(self, first, second):
         self.first = first
         self.second = second

    def setdata(self, first, second):
        self.first = first
        self.second = second

    def add(self):
        result = self.first + self.second
        return result
    def mul(self):
         result = self.first * self.second
         return result
    def sub(self):
         result = self.first - self.second
         return result
    def div(self):
         result = self.first / self.second
         return result
```

```python
a = FourCal()
b = FourCal()
a.setdata(4, 2)
b.setdata(3, 8)
print(a.add())
print(a.mul())
print(a.sub())
print(a.div())
print(b.add())
print(b.mul())
print(b.sub())
print(b.div())
print(a.first)
print(a.second)
print(b.first)
print(b.second)
```

```python
class MoreFourCal(FourCal):
    def pow(self):
        result = self.first ** self.second
        return result
```

```python
a = MoreFourCal(4, 2)
print(a.pow())
print(a.add())
```

```python
class SafeFourCal(FourCal):
    def div(self):
        if self.second == 0:  # 나누는 값이 0인 경우 0을 반환하도록 수정
            return 0
        else:
            return self.first / self.second
```

```python
a = SafeFourCal(4, 0)
print(a.div())
```

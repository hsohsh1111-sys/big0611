```python
import numpy as np
```

```python
arr = np.arange(16)
print(arr)
```

```python
arr2 = np.arange(16).reshape(4, 4)
print(arr2)
```

```python
print(arr2[1]) 
print(arr2[1][1])  # Accessing a specific element
```

```python
arr2[1, 2] = 100
print(arr2)
```

```python
print(arr2[0:2, 2:4])
print(arr2[:2, 2:])  # Slicing the array
```
